```python
import pandas as pd
import numpy as np
import seaborn as sns

# Load built-in Tips dataset (Restaurant Sales - always works!)
df = sns.load_dataset('tips')

# Show first 5 rows
df.head()
```

```python
# How many rows and columns?
print("Shape of data:", df.shape)

# Column names
print("\nColumns:", df.columns.tolist())

# Data types of each column
print("\nData Types:")
print(df.dtypes)

# Basic statistics
print("\nBasic Statistics:")
df.describe()
```

```python
# Check for missing values
print("Missing Values in each column:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# If any duplicates, remove them
df = df.drop_duplicates()
print("Data cleaned! Final shape:", df.shape)
```

```python
# Average tip by day
print("Average Tip by Day:")
print(df.groupby('day')['tip'].mean().round(2))

# Average bill by time (Lunch vs Dinner)
print("\nAverage Bill - Lunch vs Dinner:")
print(df.groupby('time')['total_bill'].mean().round(2))

# Average tip by gender
print("\nAverage Tip by Gender:")
print(df.groupby('sex')['tip'].mean().round(2))

# Who tips more - Smokers or Non-smokers?
print("\nAverage Tip - Smokers vs Non-Smokers:")
print(df.groupby('smoker')['tip'].mean().round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
plt.style.use('seaborn-v0_8')
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Restaurant Sales Analysis', fontsize=16, fontweight='bold')

# Chart 1: Tips by Day
df.groupby('day')['tip'].mean().plot(kind='bar', ax=axes[0,0], color='skyblue')
axes[0,0].set_title('Average Tip by Day')
axes[0,0].set_xlabel('Day')
axes[0,0].set_ylabel('Average Tip ($)')

# Chart 2: Bill Distribution
axes[0,1].hist(df['total_bill'], bins=20, color='salmon', edgecolor='black')
axes[0,1].set_title('Bill Amount Distribution')
axes[0,1].set_xlabel('Total Bill ($)')

# Chart 3: Lunch vs Dinner
df.groupby('time')['total_bill'].mean().plot(kind='bar', ax=axes[1,0], color='lightgreen')
axes[1,0].set_title('Average Bill: Lunch vs Dinner')
axes[1,0].set_xlabel('Time')

# Chart 4: Tip vs Bill (Scatter)
axes[1,1].scatter(df['total_bill'], df['tip'], alpha=0.6, color='purple')
axes[1,1].set_title('Tip vs Total Bill')
axes[1,1].set_xlabel('Total Bill ($)')
axes[1,1].set_ylabel('Tip ($)')

plt.tight_layout()
plt.show()
```
