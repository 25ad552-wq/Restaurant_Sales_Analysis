```python
# Import all libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset('tips')

print("=" * 40)
print("   EXPLORATORY DATA ANALYSIS (EDA)")
print("   Dataset: Restaurant Tips")
print("=" * 40)

print(f"\n Total Records : {df.shape[0]}")
print(f" Total Columns : {df.shape[1]}")
print(f"\n Column Names  : {list(df.columns)}")

print("\n First 5 rows:")
print(df.head())

print("\n Missing Values:")
print(df.isnull().sum())

print("\n Data Types:")
print(df.dtypes)
```

```python
print("=" * 40)
print("   STATISTICAL SUMMARY")
print("=" * 40)

# Detailed statistics
print(df.describe().round(2))

# Tip percentage column (new insight!)
df['tip_percent'] = (df['tip'] / df['total_bill'] * 100).round(2)
print(f"\n Average Tip Percentage : {df['tip_percent'].mean():.2f}%")
print(f" Highest Tip Percentage : {df['tip_percent'].max():.2f}%")
print(f" Lowest Tip Percentage  : {df['tip_percent'].min():.2f}%")

print("\n" + "=" * 40)
print("   GROUP ANALYSIS")
print("=" * 40)

print("\n Average tip by Day:")
print(df.groupby('day')['tip'].mean().round(2))

print("\n Average tip by Gender:")
print(df.groupby('sex')['tip'].mean().round(2))

print("\n Average bill by Time:")
print(df.groupby('time')['total_bill'].mean().round(2))

print("\n Average tip % by Smoker:")|

print(df.groupby('smoker')['tip_percent'].mean().round(2))
```

```python
fig, axes = plt.subplots(3, 2, figsize=(14, 16))
fig.suptitle('EDA - Restaurant Tips Analysis', fontsize=18, fontweight='bold')

# Chart 1: Distribution of Total Bill
axes[0,0].hist(df['total_bill'], bins=25, color='steelblue', edgecolor='white')
axes[0,0].set_title('Distribution of Total Bill')
axes[0,0].set_xlabel('Total Bill ($)')

# Chart 2: Distribution of Tips
axes[0,1].hist(df['tip'], bins=25, color='coral', edgecolor='white')
axes[0,1].set_title('Distribution of Tips')
axes[0,1].set_xlabel('Tip Amount ($)')

# Chart 3: Tips by Day (Boxplot)
df.boxplot(column='tip', by='day', ax=axes[1,0],
           boxprops=dict(color='blue'))
axes[1,0].set_title('Tip Distribution by Day')
axes[1,0].set_xlabel('Day')

# Chart 4: Bill vs Tip colored by Gender
colors = df['sex'].map({'Male': 'blue', 'Female': 'red'})
axes[1,1].scatter(df['total_bill'], df['tip'], c=colors, alpha=0.6)
axes[1,1].set_title('Bill vs Tip (Blue=Male, Red=Female)')
axes[1,1].set_xlabel('Total Bill ($)')
axes[1,1].set_ylabel('Tip ($)')

# Chart 5: Average Tip by Day and Time
df.groupby(['day','time'])['tip'].mean().unstack().plot(
    kind='bar', ax=axes[2,0], color=['orange','purple'])
axes[2,0].set_title('Avg Tip by Day and Time')
axes[2,0].set_xlabel('Day')

# Chart 6: Tip Percentage Distribution
axes[2,1].hist(df['tip_percent'], bins=25, color='green', edgecolor='white')
axes[2,1].set_title('Tip Percentage Distribution')
axes[2,1].set_xlabel('Tip %')

plt.tight_layout()
plt.show()
```
